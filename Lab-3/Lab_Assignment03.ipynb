```python
#Part-1

import random

scorpion = -1
subzero = 1
draw = 0

def ab_prun(depth, player):
  if depth == 0:
    return evaluate()
  if player == 0:
    value = float('inf')
    for m in range(2):
      value = min(value, ab_prun(depth - 1, 1))
    return value
  else:
    value = float('-inf')
    for n in range(2):
      value = max(value, ab_prun(depth - 1, 0))
  return value

def evaluate():
  return random.choice([scorpion, subzero])

def simulate_battle(starter):
  rounds = 0
  round_winners = []

  while rounds < 3:
    rounds += 1
    score = ab_prun(5, starter)

    if score == scorpion:
      round_winners.append("Scorpion")
    elif score == subzero:
      round_winners.append("Sub-Zero")
    else:
      round_winners.append("Draw")
    starter = 1 - starter
  return round_winners

input_file = open('/content/sample_data/input.txt', 'r')
output_file = open('/content/sample_data/output.txt', 'w')

starter = int(input_file.read().strip())
round_winners = simulate_battle(starter)

scorpion_wins = round_winners.count("Scorpion")
subzero_wins = round_winners.count("Sub-Zero")
draws = round_winners.count("Draw")

if scorpion_wins > subzero_wins:
  game_winner = "Scorpion"
elif subzero_wins > scorpion_wins:
  game_winner = "Sub-Zero"
else:
  game_winner = "Draw"

output_file.write(f"Game Winner: {game_winner}\n")
output_file.write(f"Total Rounds Played: {len(round_winners)}\n")
for i, winner in enumerate(round_winners):
    output_file.write(f"Winner of Round {i + 1}: {winner}\n")

input_file.close()
output_file.close()
```
