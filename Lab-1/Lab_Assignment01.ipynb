```python
input_file = open('/content/sample_data/input.txt','r')
output_file = open('/content/sample_data/output.txt','w')

import heapq



path_distances = {}
heaur_val = {}
nearest = {}

for l in input_file:
  data = l.split()
  h,n1 = int(data[1]),data[0]
  heaur_val[n1] = h
  for j in range(2,len(data),2):
    neighbour,dist = data[j], int(data[j+1])
    nearest[neighbour] = dist
  path_distances[n1] = nearest

def aster_search(start,goal):
  exp = set()
  front = [(heaur_val[start],0,start,[])]
  while front:
    front.sort()
    total_c,path_c,curr_n,path = heapq.heappop(front)
    if curr_n == goal:
      return path+[curr_n]
    if curr_n in exp:
      continue
    exp.add(curr_n)
    for nei,di in path_distances[curr_n].items():
      if nei not in exp:
        new = path+[curr_n]
        new_c = path_c + di
        total_c = new_c + heaur_val[nei]
        heapq.heappush(front,(total_c,new_c,nei,new))
  return 'No Path Found'



starting_node = input('Enter your starting node:- ')
ending_node = input('Enter your ending node:- ')

opp = aster_search(starting_node,ending_node)

if opp == 'No Path Found':
  output_file.write('No path found')
else:
  output_file.write('Optimal path: '+' -> '.join(opp) + '\n')
  total_dist = 0
  for i in range(len(opp)-1):
    current = opp[i]
    next = opp[i+1]
    distance = path_distances[current][next]
    total_dist += distance
  output_file.write('Total distance:- ' + str(total_dist) + 'km\n')
```
