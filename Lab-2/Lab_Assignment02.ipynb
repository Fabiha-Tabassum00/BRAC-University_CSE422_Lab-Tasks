```python
#Part-1

import random

mut_rate = 0.1
highest_iterate = 1000

input_file = open('/content/sample_data/input.txt', 'r')
N, T = map(int, input_file.readline().split())
courses = [input_file.readline().strip() for _ in range(N)]
array_size = N*T

def population_generator(N, T, array_size):
    arr = []
    for _ in range(array_size):
      chromosome = ''.join([str(random.randint(0, 1)) for _ in range(N * T)])
      arr.append(chromosome)
    return arr

def calculate_fitness(chromosome):
    fitness = 0
    overlap_penalty = 0
    consistency_penalty = 0

    for i in range(T):
      course_present = chromosome[i*N : (i+1)*N]
      for c in course_present:
        overlap_penalty += sum([int(c)]) - 1

    for j in range(N):
      for i in range(T):
        cnt = sum([int(chromosome[i*N + j])])
      if cnt != 1:
        consistency_penalty += 1

    total_penalty = overlap_penalty + consistency_penalty
    fitness = -total_penalty

    return fitness

def select_parents(arr):
    pnt1, pnt2 = random.sample(arr, 2)
    return pnt1, pnt2

def single_point_crossover(pnt1, pnt2):
    cross = random.randint(1, len(pnt1) - 1)
    c_1 = pnt1[:cross] + pnt2[cross:]
    c_2 = pnt2[:cross] + pnt1[cross:]
    return c_1, c_2

def mutate(chromosome, mut_rate):
    mut_chromosome = ''
    for g in chromosome:
      if random.random() < mut_rate:
          mut_chromosome += '1' if g == '0' else '0'
      else:
          mut_chromosome += g

    return mut_chromosome


arr = population_generator(N, T, array_size)

fittest = None
high_fitness = float('inf')
for _ in range(highest_iterate):
  for chromosome in arr:
    fitness = calculate_fitness(chromosome)
    if fitness < high_fitness:
      fittest = chromosome
      high_fitness = fitness

    new_pop = []
    for _ in range(array_size // 2):
      pnt1, pnt2 = select_parents(arr)
      c_1, c_2 = single_point_crossover(pnt1, pnt2)
      c_1 = mutate(c_1, mut_rate)
      c_2 = mutate(c_2, mut_rate)
      new_pop.extend([c_1, c_2])

    arr = new_pop

output_file = open('/content/sample_data/output.txt', 'w')
output_file.write(fittest+ '\n')
output_file.write(str(-high_fitness) + '\n')

input_file.close()
output_file.close()
```
